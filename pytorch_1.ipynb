```python
import torch
import time

# Define the size of the matrices
size = 10000   #Large size for performance comparison

# Create two random matrices on the CPU
matrix_cpu1 = torch.randn(size, size)
matrix_cpu2 = torch.randn(size, size)

# Measure the time taken to perform matrix multiplication on the CPU
start_time_cpu = time.time()
result_cpu = torch.matmul(matrix_cpu1, matrix_cpu2)
cpu_time = time.time() - start_time_cpu

print(f"Time taken for matrix multiplication on CPU: {cpu_time:.6f} seconds")

# Move matrices to GPU
matrix_gpu1 = matrix_cpu1.to('cuda')
matrix_gpu2 = matrix_cpu2.to('cuda')

# Measure time on GPU
start_time_gpu = time.time()
result_gpu = torch.matmul(matrix_gpu1, matrix_gpu2)
torch.cuda.synchronize()     # Ensure all GPU operations are complete
gpu_time = time.time() - start_time_gpu

print(f"Time taken for matrix multiplication on GPU: {gpu_time:.6f} seconds")


# Compare results
print("Speed : ", cpu_time / gpu_time)
```
